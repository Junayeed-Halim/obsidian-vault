# Achieve K-anonymity


K-anonymity: Each released record should be indistinguishable from at least (k-1) others on its QI attributes. 


1. Import a table or construct a table

```python
data = {'Name': ['Alice', 'Betty', 'Charles', 'David', 'Emily', 'Fred',
                     'Tom', 'Alex', 'Vincent', 'Serena'],
       'Zipcode': [47677, 47602, 47678, 47905, 47909, 47906,
                  47433,47489,47422,47443],
       'Age': [29, 22, 27, 43, 42, 47,
              32,33,34,38],
       'Sex': ['F', 'F', 'M', 'M', 'F', 'M',
              'M', 'M', 'M', 'F'],
       'Disease': ['Ovarian Cancer', 'Ovarian Cancer', 
                   'Prostate Cancer', 'Flu', 'Heart Disease', 'Heart Disease',
                  'Flu', 'Flu', 'Flu', 'Heart Disease']}
```

```python
df = pd.DataFrame(data)
df
```

Goal: release a table with 6 records that achieves 3-anonymity

2. Remove key attribute

Key attributes: Name, address, phone number- uniquely identifying!


```python
del df['Name']
df
```

3. Obscure quasi-identifier

quasi-identifier: 5-digit ZIP code, birth date, gender, Can be used for linking anonymized dataset with other datasets.

```python
# obscure zipcode to 476**
df['Zipcode'].replace([47677,47602,47678], '476**',inplace=True)
df
```

```python
# obscure zipcode to 479**
df['Zipcode'].replace([47905,47909,47906], '479**',inplace=True)
df
```

```python
# obscure zipcode to 474**
df['Zipcode'].replace([47433,47489,47422,47443], '474**',inplace=True)
df
```

```python
# obscure age to 2*
df['Age'].replace([29,22,27], '2*',inplace=True)
df
```

```python
# obscure age to 4*
df['Age'].replace([43,42,47], '4*',inplace=True)
df
```

```python
# obscure age to 3*
df['Age'].replace([32,33,34,38], '3*',inplace=True)
df
```

```python
# obscure sex to *
df['Sex'].replace(['F','M'], '*',inplace=True)
df
```

4. Release the table with 6 records that achieves 3-anonymity

```python
release_df_1 = df[:6]
release_df_1
```

# Achieve l-diversity

l-diversity:

An equivalent class is  l-diverse if contains at least l ‘well represented” values for the sensitive attribute S. A table is  l-diverse if every equivalent class is  l-diverse.


Goal: Release the table with 9 records that achieves 3-anonymity and 2-diversity

# Achieve t-closeness

What is t-closeness?

An equivalence class is said to have t-closeness if the distance between the distribution of 
a sensitive attribute in this class and the distribution of the attribute in the whole table 
is no more than a threshold t. 

The distance between two distributions commonly is calculated by the wasserstein distance/earth mover’s distance (EMD).


wasserstein_distance

Wasserstein distances arise from the so-called Optimal Transport theory. 
It aims to provide a special way to compare probability distributions.

1. Import library of wasserstein_distance

```python
from scipy.stats import wasserstein_distance
```

2. Express the disease with numbers

1: Ovarian Cancer
    
2: Prostate Cancer
    
3: Flu
    
4: Heart Disease
    

3. Calculate the distribution of the each class

```python
p_1 = [1,1,2]

p_2 = [3,4,4]

p_3 = [3,3,4]
```

4. Calculate the distribution in the whole table

```python
q = [1,1,2,3,4,4,3,3,4]
```

5. Calculate the wasserstein_distance between class_1, class_2, class_3 and the whole table.

```python
d_1 = wasserstein_distance(p_1,q)
print(d_1)

d_2 = wasserstein_distance(p_2,q)
print(d_2)

d_3 = wasserstein_distance(p_3,q)
print(d_3)
```

6. Calculate the t-closeness

```python
t = max(d_1,d_2,d_3)
print(t)
```

# summary

K-anonymity: Each released record should be indistinguishable from at least (k-1) others on its QI attributes. 

l-Diversity: Sensitive attributes must be “ l-diverse” within each quasi-identifier equivalence class.

t-closeness : t-closeness requires that the distribution of a sensitive attribute in any eq. class is close to the distribution of a sensitive attribute in the overall table.
