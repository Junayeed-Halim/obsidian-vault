# Basic use of Pandas

## What is Pandas?

**pandas** is a Python package providing fast, flexible, and expressive data structures designed to make working with "relational" or "labeled" data both easy and intuitive. It aims to be the fundamental high-level building block for doing practical, **real world** data analysis in Python. Additionally, it has the broader goal of becoming **the most powerful and flexible open source data analysis / manipulation tool available in any language**. It is already well on its way toward this goal.

# Differential Attack

Now we will implement differential attack using the example introduced in the lecture. Firstly, we create a raw dataset without any anonymity.



```python
data_da = {"Name" : ["Alice", "Bob", "Chris", "David", "Eric", "Fiona", "Grace"],
           "Grade" : [85  , 62 , 70, 93 , 58, 81 , 74]}
```

```python
df8 = pd.DataFrame(data_da)
df8
```

When we consider the attribute 'Grade' is sensitive, the dataset is anonymised as following:

```python
data_da1 = {"Name" : ["Alice", "Bob", "Chris", "David", "Eric", "Fiona", "Grace"],
"Grade" : ["**", "**", "**", "**", "**", "**", "**"]}
```

```python
df9 = pd.DataFrame(data_da1)
df9
```

However, we can still query about the raw dataset, for example, we want to know the sum of all students' grade, it will return the result to us.

```python
df8['Grade'].sum()
```

We know that the sum of all students' grade is 523. Now, we will query about the sum of the first 6 students' grade except Grace, it will retuern the result to us.

```python
df8['Grade'].sum() - df8['Grade'][6]
```

Now, we can use the first query answer to subtract the second query answer to get Grace's grade

```python
523 - 449
```

It returns Grace's grade is 74.
